```python
import langchain
langchain.__version__
```

```python
import os
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
load_dotenv()

os.environ["GEMINI_API_KEY"]=os.getenv("GEMINI_API_KEY")

llm = ChatGoogleGenerativeAI(model="gemini-3.1-flash-lite")
```

```python
from langchain.agents import create_agent
from langchain_community.tools import DuckDuckGoSearchRun

search_tool = DuckDuckGoSearchRun()

def get_weather(city:str)-> str:
    """Get the weather for a city."""
    return f"The weather in {city} is sunny."

agent=create_agent(
    model=llm,
    tools=[search_tool, get_weather],
    system_prompt="You are an AI assistant named Blox. Report the answers precisely when asked a question. You may speak freely during normal communications",
)
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt_template = ChatPromptTemplate.from_messages([
    ("human", "What are the current Weather condition like temperature, wind, rainfall chance etc etc in {place}")
])
prompt=prompt_template.invoke({"place" : "NewYork"})

response=agent.invoke({"messages" : prompt.to_messages()})
```
